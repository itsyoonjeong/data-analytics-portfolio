Every year, American high school students take SATs, standardized tests intended to measure literacy, numeracy, and writing skills.
There are three sections - reading, math, and writing, each with a **maximum score of 800 points**.

```python
import pandas as pd
schools = pd.read_csv("schools.csv")

schools.head()
```

Q1. Which NYC schools have the best math results?

```python
best_math_schools = schools[schools["average_math"] >= (0.8 * 800)][["school_name", "average_math"]].sort_values("average_math", ascending = False)

print(best_math_schools)
```

Q2. What are the top 10 performing schools based on the combined SAT scores?

```python
schools["total_SAT"] = schools["average_math"] + schools["average_reading"] + schools["average_writing"]

top_10_schools = schools[["school_name", "total_SAT"]].sort_values("total_SAT", ascending = False).head(10)

print(top_10_schools)
```

Q3. Which single borough has the largest standard deviation in the combined SAT score?

```python
# Calculate count, mean, std of total_SAT for each borough
boroughs = schools.groupby("borough")["total_SAT"].agg(["count", "mean", "std"]).round(2)
print(boroughs)
```

```python
# Filter for max std
largest_std_dev = boroughs[boroughs["std"] == boroughs["std"].max()]
print(largest_std_dev)
```

```python
# Rename the columns for clarity
largest_std_dev = largest_std_dev.rename(columns = {"count":"num_schools", "mean":"average_SAT", "std":"std_SAT"})

# Move borough from index to column
largest_std_dev.reset_index(inplace = True)

print(largest_std_dev)
```
